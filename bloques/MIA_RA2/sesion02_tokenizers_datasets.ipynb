## 1 · Tokenizar: texto -> números

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("distilbert-base-uncased")
print(tokenizer("Hola, mundo"))
```

## 2 · Subtokens

Las palabras que no están en el vocabulario se parten en subtokens (prefijo `##`).

```python
print(tokenizer.tokenize("tokenización"))
print(tokenizer.tokenize("Esta palabra no existe: zzzzz"))
```

## 3 · `attention_mask`, padding y truncation

```python
salida = tokenizer("Hola", padding=True, truncation=True, return_tensors="pt")
print(salida)
```

## 4 · Cargar un dataset con `datasets`

```python
from datasets import load_dataset

dataset = load_dataset("imdb")   # reseñas de películas: train + test
print(dataset)
```

```python
print(dataset["train"][0])       # primer ejemplo: texto + etiqueta
```

## 5 · Tokenizar todo el dataset con `.map()`

```python
def tokenizar(ejemplo):
    return tokenizer(ejemplo["text"], truncation=True, padding="max_length")

tokenizado = dataset.map(tokenizar, batched=True)
print(tokenizado["train"][0].keys())
```

## 6 · Subconjuntos para ir rápido

`imdb` tiene ~25 000 ejemplos por split. Para Colab gratuito, reduce el dataset.

```python
pequeno = dataset["train"].shuffle(seed=42).select(range(1000))
print(len(pequeno))
```
