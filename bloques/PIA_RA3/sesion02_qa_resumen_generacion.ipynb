## 1 · Question answering (extractiva)

```python
from transformers import pipeline

qa = pipeline("question-answering")

contexto = "Nuestro horario de atención al cliente es de 9 a 18 horas, de lunes a viernes."
pregunta = "¿Cuál es el horario de atención?"

print(qa(question=pregunta, context=contexto))
```

## 2 · Resumen

```python
resumidor = pipeline("summarization")

texto = ("El cliente contactó con soporte porque su pedido llegó con retraso. "
         "Tras revisar el caso, se detectó un error en la dirección de envío. "
         "Se contactó con el cliente, se corrigió la dirección y se reenvió el paquete. "
         "El cliente confirmó la recepción dos días después y quedó satisfecho.")

print(resumidor(texto, max_length=40, min_length=10))
```

## 3 · Generación de texto

Controla la longitud con `max_new_tokens` y la creatividad con `temperature`.

```python
gen = pipeline("text-generation")

print(gen("Escribe una frase de bienvenida para una tienda online:", max_new_tokens=25)[0]["generated_text"])
```

```python
# Más creatividad (temperature alta)
print(gen("Escribe una frase de bienvenida para una tienda online:", max_new_tokens=25,
          do_sample=True, temperature=1.5)[0]["generated_text"])
```
