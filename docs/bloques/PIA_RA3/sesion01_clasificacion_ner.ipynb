# Sesión 01 · Clasificación de texto y NER

SentimentLab recibe reseñas y quiere extraer entidades de textos de clientes y noticias.

Objetivos:
1. Clasificar el sentimiento de reseñas.
2. Detectar entidades (NER) en un texto.
3. Agrupar entidades compuestas.

## 1 · Clasificación de texto

```python
from transformers import pipeline

clasificador = pipeline("sentiment-analysis")

reseñas = [
    "El pedido llegó rápido y en perfecto estado.",
    "No me gustó nada, lo devolví.",
    "Cumple su función, sin más.",
]
for r in reseñas:
    print(r, "->", clasificador(r)[0])
```

## 2 · NER: reconocimiento de entidades

```python
ner = pipeline("ner")
texto = "Tim Cook presentó el nuevo iPhone en Cupertino."
print(ner(texto))
```

## 3 · Agrupar entidades compuestas

Sin `grouped_entities`, los nombres se parten en subtokens. Con `True`, se agrupan.

```python
ner_agrupado = pipeline("ner", grouped_entities=True)
print(ner_agrupado(texto))
```
