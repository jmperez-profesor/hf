## 1 · Explorar el Hub desde Python

`huggingface_hub` permite consultar el catálogo sin salir del cuaderno.

````python
from huggingface_hub import HfApi

api = HfApi()

# Modelos de análisis de sentimiento más descargados
modelos = api.list_models(task="text-classification", sort="downloads", direction=-1, limit=5)
for m in modelos:
    print(m.modelId, "| descargas:", m.downloads)
````

## 2 · La ficha de caracterización

Para cada modelo, rellena:

```
Modelo:  <usuario/modelo>
Tarea:   <qué hace>
Entrada: <tipo>  ->  Salida: <tipo>
Datos:   <dataset(s), idioma>
Métrica: <cómo se evalúa, valor>
Límites: <cuándo NO usarlo>
Licencia:<condiciones>
```

> Consulta la *model card* de cada modelo en `https://huggingface.co/<usuario>/<modelo>`.

## 3 · Probar un modelo para verificar la ficha

Caracterizar no es solo leer: **prueba** el modelo y comprueba que hace lo que dice su ficha.

````python
from transformers import pipeline

clasificador = pipeline("sentiment-analysis", model="nlptown/bert-base-multilingual-uncased-sentiment")
print(clasificador("El producto llegó roto, muy decepcionado."))
print(clasificador("Excelente, repetiré seguro."))
````
