# Sesión 01 · Pipelines y modelos preentrenados

**SentimentLab** es una tienda online que quiere medir la satisfacción de sus clientes leyendo sus reseñas. En esta sesión clasificamos reseñas en **positiva/negativa** usando un modelo preentrenado del Hub, **sin entrenar nada**.

Objetivos:
1. Usar una `pipeline` de `transformers`.
2. Entender qué devuelve (`label` y `score`).
3. Probar otras tareas y elegir modelo en el Hub.

```python
# Instala transformers (solo la primera vez en Colab)
%pip install -q transformers
```

```python
from transformers import pipeline

# Una pipeline de análisis de sentimiento en 1 línea
clasificador = pipeline("sentiment-analysis")

print(clasificador("Me ha encantado el producto, llegó antes de tiempo."))
```

## Leer el resultado: `label` y `score`

- `label`: la etiqueta predicha (`POSITIVE` / `NEGATIVE`).
- `score`: la confianza del modelo (0–1). **No** es una probabilidad calibrada.

```python
reseñas = [
    "El pedido llegó rápido y en perfecto estado.",
    "No me gustó nada, lo devolví.",
    "Bueno, cumple su función, nada del otro mundo.",
]

for r in reseñas:
    print(r, "->", clasificador(r)[0])
```

## Otras tareas con una sola llamada

Cambia solo el primer argumento de `pipeline`.

```python
# Reconocimiento de entidades (NER)
ner = pipeline("ner")
print(ner("Apple abrió una tienda en Valencia el lunes."))
```

```python
# Resumen
resumen = pipeline("summarization")
texto = ("Hugging Face es una plataforma que centraliza modelos de aprendizaje automático. "
         "Su librería transformers permite usar modelos preentrenados con pocas líneas. "
         "Gracias al transfer learning, no hace falta entrenar desde cero.")
print(resumen(texto, max_length=30, min_length=10))
```

## Elegir el modelo en el Hub

`pipeline("tarea")` usa un modelo por defecto (genérico, en inglés). Elige uno **afinado** a tu tarea e idioma:

```python
clasificador = pipeline(
    "sentiment-analysis",
    model="nlptown/bert-base-multilingual-uncased-sentiment",  # ejemplo multilingüe
)
print(clasificador("Me ha encantado el producto."))
```
