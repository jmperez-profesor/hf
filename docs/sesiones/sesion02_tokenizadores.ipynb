## 1 · Tokenizar y leer los campos

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("distilbert-base-uncased")
print(tokenizer("Hola, mundo"))
print("subpalabras:", tokenizer.tokenize("tokenización"))
```

## 2 · padding, truncación y tensores

```python
print(tokenizer("Hola", padding=True, truncation=True, return_tensors="pt"))
```

## Práctica: compara tres modelos

Tokeniza el mismo texto con BERT (WordPiece), GPT-2 (BPE) y T5 (SentencePiece) y compara el número de tokens y los cortes.

```python
texto = "La inteligencia artificial está transformando la educación y la medicina."

for nombre in ["distilbert-base-uncased", "gpt2", "t5-small"]:
    tk = AutoTokenizer.from_pretrained(nombre)
    ids = tk(texto)["input_ids"]
    print(f"{nombre}: {len(ids)} tokens -> {tk.tokenize(texto)}")
```
