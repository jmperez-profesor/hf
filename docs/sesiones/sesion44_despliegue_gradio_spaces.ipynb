## 1 · Una demo con Gradio (Interface)

````python
import gradio as gr
from transformers import pipeline

clasificador = pipeline("sentiment-analysis")

def clasificar(texto):
    r = clasificador(texto)[0]
    return f"{r['label']} ({r['score']:.2f})"

# En Colab, launch(share=True) genera un enlace público temporal
demo = gr.Interface(fn=clasificar, inputs="text", outputs="text")
demo.launch(share=True)
````

## 2 · Inferencia remota con InferenceClient (sin descargar el modelo)

````python
from huggingface_hub import InferenceClient

client = InferenceClient()
print(client.text_classification("Me encanta este producto",
      model="distilbert-base-uncased-finetuned-sst-2-english"))
````

## 3 · Publicar en un Space

Para publicar: sube un `app.py` (Gradio) y `requirements.txt` a un Space desde el Hub (o con `huggingface_hub`). Ejemplo de `app.py`:

```python
import gradio as gr
from transformers import pipeline

clasificador = pipeline("sentiment-analysis")

def clasificar(texto):
    return clasificador(texto)[0]["label"]

gr.Interface(fn=clasificar, inputs="text", outputs="text").launch()
```

## 2 · Gradio con Blocks (flujo y disposición)

````python
import gradio as gr

def traducir(texto):
    # TODO: sustituye por tu pipeline de traducción (Helsinki-NLP/opus-mt)
    return texto.upper()

with gr.Blocks() as demo:
    gr.Markdown("## Traductor")
    with gr.Row():
        entrada = gr.Textbox(label="Entrada")
        salida = gr.Textbox(label="Salida")
    btn = gr.Button("Traducir")
    btn.click(fn=traducir, inputs=entrada, outputs=salida)

demo.launch(share=True)
````
