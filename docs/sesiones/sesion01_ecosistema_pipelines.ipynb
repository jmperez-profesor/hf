## 1 · Una pipeline de sentimiento

```python
from transformers import pipeline

clasificador = pipeline("sentiment-analysis")
print(clasificador("Me ha encantado el producto, llegó antes de tiempo."))
```

## 2 · Varias tareas con la misma interfaz

```python
ner = pipeline("ner")
print(ner("Apple abrió una tienda en Valencia."))
```

```python
zero_shot = pipeline("zero-shot-classification")
print(zero_shot("Quiero devolver un producto roto", ["reclamacion", "compra", "consulta"]))
```

## 3 · Elegir modelo (multilingüe vs. español)

El modelo por defecto es genérico y en inglés. Elige uno afinado a tu idioma.

```python
multilingue = pipeline("sentiment-analysis", model="nlptown/bert-base-multilingual-uncased-sentiment")
print(multilingue("El pedido llegó roto, muy decepcionado."))
```
