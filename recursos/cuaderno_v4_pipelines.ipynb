# Cuaderno v4 · Pipelines clásicas (transformers<5)

Este cuaderno repasa las **pipelines clásicas** tal como se vieron en clase,
con `transformers` **v4**. Varias de ellas (`translation`, `summarization`,
`question-answering`) **se eliminaron en v5**; están marcadas con ⚠️.

> Para la versión con Gradio y Hub, ver el [Cuaderno v5](cuaderno_v5_gradio.ipynb).

```python
%pip install -q "transformers<5" "huggingface-hub<1.0" torch
import transformers
print("transformers:", transformers.__version__)
```

---
## 1 · Análisis de sentimiento

```python
from transformers import pipeline

clasificador = pipeline("sentiment-analysis")
print(clasificador("Me encanta este producto, llegó antes de tiempo."))
```

---
## 2 · Reconocimiento de entidades (NER)

```python
ner = pipeline("ner", aggregation_strategy="simple")
print(ner("Apple abrió una nueva tienda en Valencia."))
```

---
## 3 · Traducción EN→ES

> ⚠️ La tarea `translation` se **eliminó en v5**.

```python
traductor = pipeline("translation", model="Helsinki-NLP/opus-mt-en-es")
print(traductor("I like to play basketball and being a geek")[0]["translation_text"])
```

---
## 4 · Resumen

> ⚠️ La tarea `summarization` se **eliminó en v5**.

```python
resumidor = pipeline("summarization", model="sshleifer/distilbart-cnn-12-6")

texto = ("Hugging Face es una plataforma que centraliza modelos de aprendizaje automático. "
         "Su librería transformers permite usar modelos preentrenados con pocas líneas. "
         "Gracias al transfer learning, no hace falta entrenar desde cero.")

print(resumidor(texto, max_length=30, min_length=10)[0]["summary_text"])
```

---
## 5 · Question answering

> ⚠️ La tarea `question-answering` se **eliminó en v5**.

```python
qa = pipeline("question-answering", model="distilbert-base-cased-distilled-squad")

contexto = "El horario de atención al cliente es de 9 a 18 horas, de lunes a viernes."
print(qa(question="¿Cuál es el horario?", context=contexto)["answer"])
```

---
## 6 · Zero-shot classification

```python
zero_shot = pipeline("zero-shot-classification")
print(zero_shot("Quiero devolver un producto que llegó roto",
               ["reclamacion", "compra", "consulta"]))
```
