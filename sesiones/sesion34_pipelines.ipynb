## 1 · Una pipeline de sentimiento

Al usar `pipeline()` sin especificar modelo, HF descarga el modelo por defecto
para esa tarea. Para reproducibilidad, fija siempre la versión con `revision=`.

```python
import pandas as pd
from transformers import pipeline

# revision= fija el commit exacto del modelo → resultados reproducibles
clasificador = pipeline(
    "text-classification",
    model="distilbert-base-uncased-finetuned-sst-2-english",
    revision="af0f99b"
)

reseña = """
I thought this was a wonderful way to spend a hot summer weekend,
sitting in the air-conditioned theater watching a light-hearted comedy.
The dialogue is witty and the characters are very likable.
"""

resultado = clasificador(reseña)
pd.DataFrame(resultado)
```

### Clasificar varios textos a la vez

```python
frases = [
    "I loved the product, it arrived ahead of schedule.",
    "Terrible quality, it broke on the second day.",
    "It was okay, nothing special but not bad either.",
]

resultados = clasificador(frases)
pd.DataFrame(resultados)
```

## 2 · Varias tareas con la misma interfaz

`pipeline()` acepta más de 20 tareas. La arquitectura interna cambia,
pero la API es siempre la misma.

```python
# NER: reconocimiento de entidades nombradas
ner = pipeline("ner", aggregation_strategy="simple")
entidades = ner("Apple abrió una nueva tienda en Valencia, España.")
pd.DataFrame(entidades)
```

```python
# Zero-shot: clasifica sin haber visto ejemplos de esas etiquetas
zero_shot = pipeline("zero-shot-classification")
resultado_zs = zero_shot(
    "Quiero devolver un producto que llegó roto",
    candidate_labels=["reclamacion", "compra", "consulta", "devolución"]
)
pd.DataFrame({"etiqueta": resultado_zs["labels"], "score": resultado_zs["scores"]})
```

```python
# Generación de texto
generador = pipeline(
    "text-generation",
    model="openai-community/gpt2"
)
salida = generador(
    "In this course, we will teach you how to",
    max_length=50,
    num_return_sequences=2,
    do_sample=True,
    temperature=0.9
)
for i, s in enumerate(salida, 1):
    print(f"\n--- Secuencia {i} ---")
    print(s["generated_text"])
```

## 3 · Elegir modelo: impacto del idioma

El modelo por defecto está entrenado en inglés. Para español necesitas uno
multilingüe o específico. La model card del Hub te dice el idioma soportado.

```python
modelo_ingles = pipeline(
    "text-classification",
    model="distilbert-base-uncased-finetuned-sst-2-english",
    revision="af0f99b"
)

modelo_multilingue = pipeline(
    "text-classification",
    model="nlptown/bert-base-multilingual-uncased-sentiment"
)

frases_esp = [
    "El pedido llegó roto, muy decepcionado.",
    "Me ha encantado el producto, funciona perfecto.",
    "Ni bien ni mal, simplemente normal.",
]

print(f"{'Frase':45s}  {'Inglés':18s}  {'Multilingüe'}")
print("-" * 85)
for frase in frases_esp:
    r_en = modelo_ingles(frase)[0]
    r_ml = modelo_multilingue(frase)[0]
    print(f"{frase[:45]:45s}  {r_en['label']:6s} {r_en['score']:.2f}  {r_ml['label']:6s} {r_ml['score']:.2f}")
```

## 4 · Gradio: envuelve tu modelo en una interfaz

Con [Gradio](https://www.gradio.app/) conviertes cualquier `pipeline` en una **app web**
interactiva, sin HTML/CSS/JS. Es la forma más rápida de "usar" un modelo sin programar.

> Lo ampliamos (componentes, disposición, `Blocks`) y lo desplegamos a **Spaces** en la sesión 34.

```python
%pip install -q gradio
import gradio as gr
from transformers import pipeline

clasificador = pipeline(
    "text-classification",
    model="nlptown/bert-base-multilingual-uncased-sentiment"
)

def analizar(texto):
    r = clasificador(texto)[0]
    return f"{r['label']} ({r['score']:.2f})"

demo = gr.Interface(
    fn=analizar,
    inputs="text",
    outputs="text",
    examples=["Me encanta este producto", "Llegó roto y tarde"]
)
demo.launch()
```

## Gradio + traducción EN→ES (compatible v4 y v5)

`pipeline("translation", ...)` se **eliminó en transformers v5**. La forma compatible
(funciona en v4 y v5) es usar `AutoModelForSeq2SeqLM` + `generate()` y envolverlo en Gradio:

> Detalle del cambio: [Migración v4 → v5](../recursos/migracion_v5.md).

```python
%pip install -q gradio
import gradio as gr
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

modelo = AutoModelForSeq2SeqLM.from_pretrained("Helsinki-NLP/opus-mt-en-es")
tokenizer = AutoTokenizer.from_pretrained("Helsinki-NLP/opus-mt-en-es")

def traducir(texto):
    entradas = tokenizer(texto, return_tensors="pt")
    salida = modelo.generate(**entradas)
    return tokenizer.decode(salida[0], skip_special_tokens=True)

gr.Interface(
    fn=traducir, inputs="text", outputs="text",
    title="Traductor EN → ES",
    examples=["I like to play basketball and being a geek"],
).launch()
```

## Gradio + chat model: resumen y pregunta-respuesta (GPU)

En v5 se eliminaron `summarization` y `question-answering`. La alternativa es un
**chat model** con `text-generation` + prompt, envuelto en Gradio.

> ⚠️ Requiere GPU (**Runtime → T4**) y descarga ~8 GB. Se reutiliza el mismo modelo
> en las dos demos. Ver [Demos con Gradio (v5)](../recursos/gradio_demos.md).

```python
%pip install -q gradio
import gradio as gr
from transformers import pipeline

pipe_llm = pipeline("text-generation", model="Qwen/Qwen3-4B-Instruct-2507")

def resumir(texto):
    r = pipe_llm([{"role": "user", "content": f"Resume en 2 frases:\n\n{texto}"}])
    return r[0]["generated_text"][-1]["content"]

gr.Interface(fn=resumir, inputs="text", outputs="text", title="Resumen").launch()
```

```python
# Reutiliza pipe_llm del bloque anterior
def responder(contexto, pregunta):
    r = pipe_llm([{"role": "user", "content":
        f"Contexto: {contexto}\n\nPregunta: {pregunta}\nRespuesta:"}])
    return r[0]["generated_text"][-1]["content"]

gr.Interface(
    fn=responder,
    inputs=[gr.Textbox(lines=5, label="Contexto"), gr.Textbox(label="Pregunta")],
    outputs="text",
    title="Pregunta-respuesta",
).launch()
```
